```python
def AND(a,b):
    return a and b 
def OR (a,b):
    return a or b
def NOT(a):
    return not a 
def XOR(a,b):
    return OR(AND(a,NOT(b)),
              AND(NOT(a),b))
    
print("Tabla de verdad usando clase:")
print("A\tB\tAND\tOR\tXOR")
print("-" * 30)
for a in [False, True]:
    for b in [False, True]:
        and_result = AND(a, b)
        or_result = OR(a, b)
        xor_result = XOR(a, b)
        
        print(f"{int(a)}\t{int(b)}\t{
            int(and_result)}\t{
                int(or_result)}\t{
                    int(xor_result)}")
    
```
