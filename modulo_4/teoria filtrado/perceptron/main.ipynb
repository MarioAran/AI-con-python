```python
class CompuertaLogica:
    @staticmethod
    def AND(a, b):
        return a and b
    @staticmethod
    def OR(a, b):
        return a or b
    @staticmethod
    def NOT(a):
        return not a
    @classmethod
    def XOR(cls, a, b):
        """Implementación de XOR usando AND, OR y NOT"""
        return cls.OR(
            cls.AND(a, cls.NOT(b)),
            cls.AND(cls.NOT(a), b)
        )
def demo_compuertas_clase():
    compuerta = CompuertaLogica()
    print("Tabla de verdad usando clase:")
    print("A\tB\tAND\tOR\tXOR")
    print("-" * 30)
    
    for a in [False, True]:
        for b in [False, True]:
            and_result = compuerta.AND(a, b)
            or_result = compuerta.OR(a, b)
            xor_result = compuerta.XOR(a, b)
            
            print(f"{int(a)}\t{int(b)}\t{
                int(and_result)}\t{
                    int(or_result)}\t{
                        int(xor_result)}")

# Ejecutar demo
demo_compuertas_clase()
```
